4. Use a função de otimização de SciPy para maximizar f(x) = −x² + 4x + 1.

```python
from scipy import optimize

# SciPy só tem função pronta para MINIMIZAR, então pra maximizar f(x) eu minimizo -f(x)
f = lambda x: -x**2 + 4*x + 1
resultado = optimize.minimize_scalar(lambda x: -f(x))

x_maximo = resultado.x
valor_maximo = f(x_maximo)

print("x que maximiza:", x_maximo)
print("valor máximo:", valor_maximo)
```

10. Calcule a integral dupla da função f(x,y) = xy + x² no intervalo [0,2] para x e [0,1] para y.

```python
from scipy import integrate

# dblquad integra em relação a y primeiro, depois a x
# a função precisa receber (y, x) nessa ordem
funcao = lambda y, x: x * y + x**2
resultado, erro_estimado = integrate.dblquad(funcao, 0, 2, 0, 1)

print("Resultado da integral dupla:", resultado)
```

16. Resolva a equação diferencial de segunda ordem d²y/dx² + y = 0, com condições iniciais y(0) = 1 e y'(0) = 0.

```python
from scipy.integrate import solve_ivp

# o solve_ivp só resolve equações de 1ª ordem, então transformo a de 2ª ordem em um SISTEMA de duas de 1ª ordem:
# y0 = y, y1 = y' -> y0' = y1  e  y1' = -y0
def sistema(x, Y):
    y, dy = Y
    return [dy, -y]

# condição inicial: y(0)=1, y'(0)=0
solucao = solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100))

plt.plot(solucao.t, solucao.y[0])
plt.xlabel("x")
plt.ylabel("y")
plt.title("Solução de y'' + y = 0 (equivale a y = cos(x))")
plt.show()
```

8. Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função y = sin(x).

```python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
margem_erro = 0.15  # intervalo de confiança fictício pra exemplo

plt.figure(figsize=(6,4))
plt.plot(x, y, color='navy', label='y = sin(x)')
# fill_between desenha a área sombreada entre (y - margem) e (y + margem)
plt.fill_between(x, y - margem_erro, y + margem_erro, alpha=0.3, color='navy')
plt.legend()
plt.title("y = sin(x) com intervalo de confiança")
plt.show()
```

18. Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

```python
categorias = ['A', 'B', 'C', 'D']
valores = [10, -5, 8, -3]

# lista de cores: verde se o valor for positivo, vermelho se for negativo
cores = ['green' if v > 0 else 'red' for v in valores]

plt.figure(figsize=(6,4))
plt.bar(categorias, valores, color=cores)
plt.axhline(0, color='black', linewidth=0.8)  # linha no zero pra facilitar a leitura
plt.title("Valores positivos e negativos")
plt.show()
```
