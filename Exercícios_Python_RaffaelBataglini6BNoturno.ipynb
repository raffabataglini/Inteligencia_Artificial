21. Busca Binária

```python
# variável numbers guaradará numeros inteiros
numbers = [23, 21, 17, 8, 5, 51, 21, 9, 12]
numbers.sort()

# variável num a ser procurado
num = 23

# variáveis das índices
inicio = 0
fim = len(numbers) - 1

# while para continuar a busca até que o início fique maior que o fim
while inicio <= fim:
    # variável ind_meio guardará a posição do meio da lista
    ind_meio = (inicio + fim) // 2

    # if-else para ajustar a índice do meio
    if num < numbers[ind_meio]:
        fim = ind_meio - 1
    elif num > numbers[ind_meio]:
        inicio = ind_meio + 1
    else:
        print(f'Número {num} encontrado no índice {ind_meio}')
        break
else:
    print(f'Número {num} não encontrado')
```

22. Criptografia de Cifra de César

```python
# variável alf guardará o alfbeto normal
alf = "abcdefghijklmnopqrstuvwxyz"

# variável cesar a criptografia de César
cesar = "defghijklmnopqrstuvwxyzabc"

# criei uma função com def chamada cript_cesar que teremos que passar um parâmetro
def cript_cesar(word):

  #  com o método lower() colocaremos tudo em minúsculo,
  word = word.lower()

  # variavel cript armazenará a mensagem criptografada
  cript = ''

  # for para interar na palavra
  for i in range(len(word)):
    # índice no alfabeto normal e que representa da criptografia
    indice = alf.index(word[i])

    # e concatenamos com a variável critp que guardará a mensagem final
    cript += cesar[indice]

  # depois a retornaremos.
  return cript

cript_cesar('cesar')
```

24. Validação de CPF


```python
# função validador_cpf criada com def com parâmtro cfp
def validador_cpf(cpf):

  # variavel guaradará o cpf e com o método replace excluirá caracteres (-,)
  cpf = cpf.replace('-','').replace('.','')

  # variável ac nosso acumulador irá guardar a soma necessária para a verificação variável cont será o contador
  ac, cont = 0,0

  # if verifica caso todos número iguais ex.(111.111.111-11)
  if len(set(cpf)) == 1:
    print("CPF inválido: dígitos repetidos")
    return False

  # if verifica se a quantidade de digitos bate
  if len(cpf) != 11:
    print("CPF inválido: tamanho incorreto")
    return

  # for para a mutiplicação de cada dígito do cfp por uma contagem regressiva de 10 até 2
  for i in range(10, 1, -1):
    ac += (int(cpf[cont]) * i)
    cont += 1

  # variável d1 guardará o primeiro digito verificador
  d1 = 0 if 11 - (ac % 11) < 2 else 11 - (ac % 11)

  cont, ac = 0, 0

  for i in range(11, 1, -1):
    ac += (int(cpf[cont]) * i)
    cont += 1

  # d2 o segundo dígito verifiador
  d2 = 0 if 11 - (ac % 11) < 2 else 11 - (ac % 11)

  if cpf[9] == str(d1) and cpf[10] == str(d2):
    print("Cpf válido")
  else:
    print("Cpf inválido")

cpf = input('Digite seu CPF: ')
validador_cpf(cpf)
```

27. Anagramas

```python
# variável word1 e word2 com as palavras que iremos testar
word1 = 'amor'
word2 = 'roma'

# função anagram criada com 2 parâmetros
def anagram(word1, word2):

    # Primeiro verifica se as palavras possuem o mesmo tamanho
    if len(word1) != len(word2):
        return False

    # Ordena as letras e compara as duas palavras
    if sorted(word1) == sorted(word2):
        return True
    else:
        return False


print(anagram(word1, word2))
```

30. Jogo da Forca

```python
# variável secret_word guardará a palavra secreta
secret_word = "gato"

# variável try_words guardará uma lista de letras chutadas
try_words = []

# limite de tentativas na variável attempts
attempts = 5

# while enquanto tentativas for maior que 0
while attempts > 0:
    # variável letter guardará a entrada do usuário
    letter = input('\nDigite uma letra: ').lower()

    # verificação se a letra já foi tentada
    if letter in try_words:
        print('Letra já digitada!')
        continue

    try_words.append(letter)

    # se a letra não estiver na palavra secreta, decrementa as tentativas
    if letter not in secret_word:
        attempts -= 1
        print(f'Letra incorreta! Tentativas restantes: {attempts}')

    # for para exibir o progresso da palavra secreta
    ganhou = True
    for i in secret_word:
        if i in try_words:
            print(i, end=' ')
        else:
            print('_', end=' ')
            ganhou = False

    if ganhou:
        print('\nParabéns! Você venceu!')
        break
```

31. Ordenação por Bolha

```python
# variável numbers guardará uma lista de números
numbers = [64, 87, 10, 4, 98, 35, 12, 7]
tamanho = len(numbers)

# for para controlar o número de passagens pela lista
for i in range(tamanho):
    # for interno para comparar elementos
    for j in range(0, tamanho - i - 1):
        # if para trocar os elementos se estiverem fora de ordem
        if numbers[j] > numbers[j + 1]:
            aux = numbers[j]
            numbers[j] = numbers[j + 1]
            numbers[j + 1] = aux

print(f'Lista ordenada: {numbers}')
```

32. Jogo da velha

```python
tabuleiro = [
    ['1', '2', '3'],
    ['4', '5', '6'],
    ['7', '8', '9']
]

def mostrar_tabuleiro(tabuleiro):
  for linha in tabuleiro:
    print(" | ".join(linha))

def verificar_vitoria(tab, jogador):
  for i in range(3):
    # verificar linhas
    if tab[i][0] == jogador and tab[i][1] == jogador and tab[i][2] == jogador:
      return True

    # verificar colunas
    if tab[0][i] == jogador and tab[1][i] == jogador and tab[2][i] == jogador:
      return True

    # verificar cruzados
    if tab[0][0] == jogador and tab[1][1] == jogador and tab[2][2] == jogador:
      return True

    if tab[2][0] == jogador and tab[1][1] == jogador and tab[0][2] == jogador:
      return True

  return False

jogador = "X"
jogadas = 0

while jogadas < 9:
  mostrar_tabuleiro(tabuleiro)
  escolha = int(input(f'Jogado {jogador} escolha um posição de (1-9): '))
  pos = escolha -1
  linha, col = pos // 3, pos % 3

  if tabuleiro[linha][col] in ['X', 'O']:
    print('Posição ja ocupada, tente outra')
    continue

  tabuleiro[linha][col] = jogador
  jogadas += 1

  if verificar_vitoria(tabuleiro, jogador):
    mostrar_tabuleiro(tabuleiro)
    print(f'Jogador {jogador} venceu!')
    break

  if jogador == "O":
    jogador = "X"
  else:
    jogador = "O"

else:
    mostrar_tabuleiro(tabuleiro)
    print('Empate!')
```

37. Funções Recursivas


```python
# função recursiva para calcular o fatorial de um número
def fatorial(n):
    # caso se n for 0 ou 1, o fatorial será 1
    if n == 0 or n == 1:
        return 1
    else:
        return n * fatorial(n - 1)

# variável num guardará a entrada do usuário
num = int(input('Digite um número para o fatorial recursivo: '))
x = fatorial(num)

print(x)
```

```python
# variável numbers guardará uma lista de inteiros
numbers = [4, 9, 8, 6 , 5, 11, 2, 7, 1, 3]

# variáveis par e ímpar guardará listas vazias
par = []
impar = []

# for para iterar cada item da lista
for i in numbers:

  # if-else para verificar o resto da divi~sao por 2
  if i % 2 == 0:

    #método append() para adicionar na lista correspondente
    par.append(i)
  else:
    impar.append(i)

print(f'Números pares: {par}')
print(f'Números pares: {impar}')
```

46. Calcular Média e Desvio Padrão


```python
# variável numbers guardará uma lista de números
numbers = [145, 150, 147, 152, 146]

tamanho = len(numbers)
media = sum(numbers)/tamanho
ac = 0

# for para calcular a soma dos quadrados das diferenças
for i in numbers:
  desvio = (i - media) **2
  ac += desvio

# variáveis de variância e desvio padrão
variancia = ac/ (tamanho-1)
desvio_padrao = variancia ** 0.5

print(f'Média: {media}')
print(f'Desvio padrão: {desvio_padrao:.2f}')
```

48. Ordenação por Inserção

```python
# numbers guardará um lista de números
numbers = [12, 11, 13, 5, 6]

# for a partir do segundo elemento da lista
for i in range(1, len(numbers)):
    chave = numbers[i]
    j = i - 1

    # move os elementos maiores que a chave para uma posição à frente
    while j >= 0 and chave < numbers[j]:
        numbers[j + 1] = numbers[j]
        j -= 1
    numbers[j + 1] = chave

print(numbers)
```

50. Cálculo de Frequência de Letras

```python
# variável texto armazenará a entrada do usuário, já o coland tudo maiúsculo
texto = input("Digite um palavra: ").upper()

# variável x guardará um dicionário vazio
x = {}

# for para percorrer cada letra da palavra
for i in texto:
  if i not in x.keys():  # se não ter adiciona
    x[i] = 1
  else:
    x[i]= x[i] + 1  # se já ter acrescenta
print(x)
```
